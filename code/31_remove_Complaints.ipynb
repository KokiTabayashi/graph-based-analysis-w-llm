### Neo4j

```python
driver = neo4j.GraphDatabase.driver(
    uri=os.environ.get("NEO4J_URI"), 
    auth=(os.environ.get("NEO4J_USERNAME"), 
          os.environ.get("NEO4J_PASSWORD"))
)
```

```python
session = driver.session(database="neo4j")
```

### Delete Complaint nodes from the knowledge graph with Category/Company nodes materialized.

```python
query = """

MATCH (n:Complaint)
DETACH DELETE n;

"""

session.run(query)
```
